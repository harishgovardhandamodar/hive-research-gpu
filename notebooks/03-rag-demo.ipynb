# RAG Demo — Hybrid Search

Compare vector, keyword, and hybrid search modes side by side.

```python
import requests
import pandas as pd
from IPython.display import display, Markdown

BASE = "http://localhost:7777"

def post(path, data):
    return requests.post(f"{BASE}{path}", json=data).json()
```

## 1. Vector Search (Semantic)

```python
query = "What are key advancements in transformer architectures?"
vec_results = post("/api/query", {"question": query, "mode": "vector"})
display(Markdown(f"**Answer:** {vec_results['answer']}"))
print("\nSources:")
for s in vec_results.get('sources', []):
    print(f"  - [{s['id']}] {s['title']}")
```

## 2. Hybrid Search (Vector + BM25)

```python
hybrid_results = post("/api/query", {"question": query, "mode": "hybrid"})
display(Markdown(f"**Answer:** {hybrid_results['answer']}"))
print("\nSources:")
for s in hybrid_results.get('sources', []):
    print(f"  - [{s['id']}] {s['title']}")
```

## 3. Ask Your Own Question

```python
question = input("Enter your question: ")
result = post("/api/query", {"question": question, "mode": "hybrid"})
display(Markdown(f"**Answer:** {result['answer']}"))
print("\nSources:")
for s in result.get('sources', []):
    print(f"  - [{s['id']}] {s['title']}")
```
