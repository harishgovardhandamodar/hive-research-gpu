# Hive Research GPU — Quick Start Notebook

This notebook demonstrates the core Python API for interacting with your research knowledge base.

## Prerequisites
Make sure the server is running: `python -m hive_research serve`

```python
import requests
import json

BASE = "http://localhost:7777"

def api(method, path, data=None):
    url = f"{BASE}{path}"
    if method == "GET":
        r = requests.get(url)
    else:
        r = requests.post(url, json=data or {})
    return r.json()
```

## 1. Check System Status

```python
stats = api("GET", "/api/stats")
print(f"Papers: {stats.get('papers', 0)}")
print(f"Concepts: {stats.get('concepts', 0)}")
print(f"Relations: {stats.get('relations', 0)}")
print(f"RAG Chunks: {stats.get('rag', {}).get('chunks', 0)}")
```

## 2. Browse Papers

```python
papers = api("GET", "/api/papers")
print(f"Total papers: {len(papers)}\n")
for p in papers[:5]:
    print(f"  [{p['id']}] {p['title'][:70]}")
    print(f"      {p.get('authors', '')[:60]}")
    print()
```

## 3. Add a Paper

```python
result = api("POST", "/api/add", {"id": "1706.03762"})
print(json.dumps(result, indent=2))
```

## 4. View Knowledge Graph

```python
graph = api("GET", "/api/graph")
nodes = graph.get("nodes", [])
links = graph.get("links", [])
print(f"Graph: {len(nodes)} nodes, {len(links)} edges")
```

## 5. Ask a Question (RAG)

```python
response = api("POST", "/api/query", {"question": "What is the transformer architecture?"})
print(response.get("answer", "No answer"))
```
